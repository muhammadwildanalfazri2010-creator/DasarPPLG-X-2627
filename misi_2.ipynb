```python


print("=== PROGRAM KONVERSI SUHU CELSIUS KE FAHRENHEIT ===")

# 1. Memasukkan suhu
celcius = float(input("Masukkan suhu dalam derajat Celcius: "))

# 2. Menghitung suhu
fahrenheit = (celcius * 1.8) + 32

# 3. Menampilkan hasil
print(f"Suhu {celcius}°C sama dengan {fahrenheit}°F")
```
